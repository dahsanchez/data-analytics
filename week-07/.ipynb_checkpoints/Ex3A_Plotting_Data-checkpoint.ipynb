# Exercise 3.A Getting familiar with Matplotlib 

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5] 
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, marker='o')
plt.show()
```

```python
plt.plot(x,y, marker='o')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

 ## Styling Plots

```python
plt.plot(x,y,'o--r')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
plt.plot(x,y, marker='s', ms=10, mfc='g')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
plt.plot(x,y, marker='*', ms=23, mfc='m',ls = 'dotted', c = 'k' )
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
plt.plot(x,y, marker='P', ms=15, mfc='c',ls = 'None',)
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

# Exercise 3.B More plotting practice 

```python
products = ['Product A', 'Product B', 'Product C', 'Product D'] 
sales = [23, 17, 35, 29] 
```

```python
plt.title('Sales by Product')
plt.xlabel('Products')
plt.ylabel('Number of Sales')
plt.bar(products,sales,
        color = 'skyblue',
        width = 0.5)
```

```python
x = np.random.rand(50) 
y = np.random.rand(50)
```

```python
plt.scatter(x,y,s = np.random.rand(50) * 100, c=np.random.rand(50),cmap='plasma')
```

## Subplots

```python
x = np.linspace(0, 10, 100) 
y = np.sin(x) 
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
data = np.random.randn(1000)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].plot(x, y, color='deeppink', linestyle='--')
axes[0].set_title('Sine Wave')
axes[0].set_xlabel('X values')
axes[0].set_ylabel('sin(X)')

axes[1].hist(data, bins=30, color='cyan', edgecolor='black')
axes[1].set_title('Histogram of Random Data')
axes[1].set_xlabel('Value')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
