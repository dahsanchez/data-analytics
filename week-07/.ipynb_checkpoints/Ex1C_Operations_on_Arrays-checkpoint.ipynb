## Sample Arrays

```python
r_1 = np.random.randint(1, 20, size=(2))
r_2 = np.random.randint(1, 20, size=(1, 2))
r_3 = np.random.rand(2, 1)
r_4 = np.random.randint(20, 100, size=(4, 5))
r_5 = np.random.rand(4,5)
```

## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

these two cells of code are in decending order

```python
new_shuffle = (np.random.permutation(r_4))
print(r_4)
print(new_shuffle)
```

the differnce between shuffle and permutation is that permutation is completely random when shuffle isnt.

## Grocery Problem

```python
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])

print("bag1:", bag1)
print("bag2:", bag2)
```

```python
bag_1_total = np.sum(bag1)
bag_2_total = np.sum(bag2)
print(f'The first bag contains {bag_1_total} items, and the second bag contains {bag_2_total} items.')
```

```python
total_bags = np.add(bag1,bag2)

print(f'There are {total_bags[0]} vegetables, {total_bags[1]} fruits, {total_bags[2]} boxes of snacks, and {total_bags[3]} beverages in the two grocery bags.')
```

## Teamwork

```python
ali = np.array([5,4,6,4,2])
```

```python
ben = np.array([0,5,5,3,4])
```

```python
carla = np.array([3,6,0,3,4])
```

```python
all_knocks = np.array([ali,ben,carla])
print(all_knocks)
```

```python
person_total = np.sum(all_knocks, axis = 1)
group_total = np.sum(all_knocks)

print(person_total)
print(group_total)
```

```python
day_total = np.sum(all_knocks, axis = 0)
votes = np.cumsum(day_total) * 0.5
print(votes)
```
