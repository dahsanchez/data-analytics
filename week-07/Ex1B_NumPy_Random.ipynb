 ## Generating Random Choices from an Array

```python
students = np.array(['Dahlia',
                     'Destiney',
                     'Brandon',
                     'Kevin',
                     'Luis',
                     'Sthefanine',
                     'Edmund',
                     'Hritik',
                     'Natalia',
                     'Jeremiah'])
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3,4)) # this isnt a good way to make groups because people can be repeated in the groups
print(breakout_groups)
```

```python
call_on_weighted = np.random.choice(students,p = [0.0, 0.3, 0.1, 0.2, 0.0, 0.1,0.0,0.1,0.1, 0.1], size = (5))
print(call_on_weighted) # Dahlia,Luis,Edmund=0%;Brandon,Sthefanine,Hritik,'Natalia,Jeremiah =10%; Destiney = 30%; Kevin = 20%
```
