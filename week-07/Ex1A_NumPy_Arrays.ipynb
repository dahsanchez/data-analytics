## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    x = x*2
    print(x)
```

```python
for x in arrA:
    for y in x:
        y = y*2
        print(y)
```

These two are differnt because the first for loop is printing the arrays as two seperate arrys , but the second for loop is printing th values in the arrays.

```python
for x in arrA:
    loop_result = x * 2
    for y in x:
        print(loop_result) 
# this is differnt then before because its looping 3 times for each number in the array.
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA))
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))

# ValueError: all input arrays must have the same shape
# we get this error because joined_array_1 is in axis 0 yet joined_array_2 is in axis 1
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis = 1)
print(another_2)

# Since joinedarray1 is bigger then arrA it can only br concatenated at axis = 0 
```

```python
harrA = np.hstack((arrA, arrA)) # stacking arrA on top of arrA
print(harrA)
darrA = np.dstack((arrA, arrA)) # stacking alomg height and depth
print(darrA)
varrA = np.vstack((arrA, arrA)) # stacking vertically on the axis of the array
print(varrA)
```

## Splitting Arrays

```python
splitsky =  np.array_split( joined_array_1, 2) # splits into two arrays both in the secound dimension
print(splitsky)
```

```python
spliterino = np.array_split( joined_array_1, 3) # there are 3 arrays all in secound dimension
print(spliterino)
```
