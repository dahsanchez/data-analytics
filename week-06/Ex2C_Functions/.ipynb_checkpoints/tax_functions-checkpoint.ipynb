## calcuating taxes with functions

```python
def get_soc_sec_tax(gross):
    gross = gross * 0.062
    return (f'The Social Security tax amount is ${format(gross,".2f")}')
```

```python
print(get_soc_sec_tax(23000))
```

```python
def get_medicare_tax(gross):
    gross = gross * 0.0145
    return (f'The Medicare tax amount is ${format(gross,".2f")}')
```

```python
print(get_medicare_tax(23000))
```

```python
def get_federal_tax(gross,with_code):
    if with_code == 0:
        gross = gross * 0.23
    elif with_code == 1:
        gross =  gross * 0.21
    elif with_code == 2:
        gross = gross * 0.195
    elif with_code == 3:
        gross = gross * 0.185
    else:
        if with_code > 4:
            dif = with_code - 4
            tax_rate = 0.18 - (dif * 0.005)
            gross = gross * tax_rate
        gross = gross * 0.18
    return (f' Federal tax reduction with your {with_code} is ${gross}.')        
```

### Person1: gross pay $750 withholding code 0 

```python
print(get_soc_sec_tax(750))
```

```python
print(get_medicare_tax(750))
```

```python
print(get_federal_tax(750,0))
```

### Person2: gross pay $1550 withholding code 2

```python
print(get_soc_sec_tax(1550))
```

```python
print(get_medicare_tax(1550))
```

```python
print(get_federal_tax(1550,2))
```
