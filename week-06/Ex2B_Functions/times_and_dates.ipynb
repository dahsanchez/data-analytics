## working with datetime functions


```python
import datetime
```

# Today

```python
def today():
    return datetime.datetime.now()
```

```python
print(today().strftime("Today is %A, %B %d, %Y"))
```

# My birthday

```python
def bday():
    return datetime.date(2003,8,4)
```

```python
print(bday().strftime('My birthday is %B %d %Y'))
```

# Next 90 Days

```python
def ninety_d():
    return today() + datetime.timedelta(days=90)
```

```python
print(ninety_d().strftime('90 days from today is %B %d %Y'))
```

# Time for Dinner

```python
def dinner_time():
    return datetime.time(18, 30)
```

```python
print(dinner_time().strftime("Let’s meet for dinner at %I:%M %p"))
```
